```python
# 시각화

import matplotlib
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'retina'

matplotlib.rcParams["font.family"] = "Malgun Gothic"   # Windows
# matplotlib.rcParams["font.family"] = "AppleGothic"   # macOS
matplotlib.rcParams["axes.unicode_minus"] = False       # 마이너스 부호 깨짐 방지
```

### 기능 1 - 정제 데이터 불러오기(load_clean_data)

```python
import pandas as pd
import os
import numpy as np

# 파일 경로
BASE_DIR = os.getcwd() # 현재 경로의 파일(상대경로 사용)
FILE_PATH = os.path.join(os.path.dirname(BASE_DIR),   # 현재 파일 경로의 dir 찾기
                         "data","student_habits_clean.csv")

# 데이터 불러오기
def load_clean_data(file_path):
    if not os.path.exists(file_path):
        print(f"과제 2를 먼저 실행하세요: {file_path}")
        return None

    df = pd.read_csv(file_path, encoding="utf-8-sig")

    print(f"데이터 로드 완료: {df.shape[0]}행 X {df.shape[1]}열")
    return df
```

### 기능 2 - 단변량 분포 시각화(plot_distributions)

```python
# 수치형 컬럼 각각의 분포를 히스토그램으로 그려 데이터가 어떻게 퍼져 있는지 확인

def plot_distribution(df):
    columns = ['sleep_hours', 'study_hours', 'phone_hours', 'exercise_hours']

    fig, axes = plt.subplots(2, 2, figsize=(12, 8)) # 2 x 2 그래프
    axes = axes.flatten()

    for i, col in enumerate(columns):
        axes[i].hist(df[col], bins=15)
        
        axes[i].set_title(f'{col}')
        axes[i].set_xlabel('hours')
        axes[i].set_ylabel('distribution')

    plt.subplots_adjust(wspace=0.3, hspace=0.3) # subplot 간에 간격 조정
    plt.tight_layout()
    
    to_path = os.path.join(os.path.dirname(BASE_DIR), 'outputs', 'distribution.png') # hist 시각화 한 파일을 저장할 경로 지정    
    plt.savefig(to_path) # hist 시각화 파일 저장
    
    plt.close(fig)
        
    print(f'\n저장완료: {os.path.relpath(to_path, BASE_DIR)}') # 저장한 파일의 상대경로 보기
```

### 기능 3 - 상관계수 표 출력 (show_correlation)

```python
# 수치형 변수 간 상관계수를 계산

def show_correlation(df):
    columns = df.select_dtypes(['int', 'float']).columns

    print('\n[상관계수 표]')
    print(df[columns].corr().round(2))
```

### 기능 4 - 이변량 분석: 산점도 (plot_scatter)


```python
# 학습 성과(점수)와 주요 변수의 관계를 산점도로 확인

def plot_scatter(df):
    columns = ['study_hours', 'sleep_hours', 'phone_hours']

    fig, axes = plt.subplots(1, 3)
    axes = axes.flatten()

    for i, col in enumerate(columns):
        axes[i].scatter(df['score'], df[col], alpha=0.5)
        
        axes[i].set_title(f'{col}, score corr')
        axes[i].set_xlabel('score')
        axes[i].set_ylabel(f'{col}')
        
    plt.subplots_adjust(wspace=0.6) # 그래프 간 간격 조정
    plt.tight_layout()
    
    to_path = os.path.join(os.path.dirname(BASE_DIR), 'outputs', 'scatter_plot.png') # scatter 시각화 한 파일을 저장할 경로 지정        
    plt.savefig(to_path) # scatter 시각화 파일 저장
    
    plt.close()
    
    print(f'\n저장완료: {os.path.relpath(to_path, BASE_DIR)}') # 저장한 파일의 상대경로 보기
```

### 기능 5 - 그룹별 비교 (plot_group_comparison)

```python
# 파생변수(sleep_sufficient, phone_overuse)로 그룹을 나눠 평균 점수를 비교

def plot_group_comparison(df):
    df_sleep = df.groupby(['sleep_sufficient'])[['score']].mean().reset_index() # sleep_sufficient를 기준으로 그룹화 후 평균 성적 내기
    df_phone = df.groupby(['phone_overuse'])[['score']].mean().reset_index() # phone_overuse를 기준으로 그룹화 후 평균 성적 내기 

    fig, axes = plt.subplots(1, 2)
    axes[0].bar(df_sleep["sleep_sufficient"], df_sleep["score"]) # 0번 그래프에 sleep 그룹화 한 데이터 생성
    axes[1].bar(df_phone["phone_overuse"], df_phone["score"]) # 1번 그래프에 phone 그룹화 한 데이터 생성


    # 그래프 제목 설정
    axes[0].set_title('수면 충족 여부에 따른 평균 성적')
    axes[1].set_title('폰 과의존 여부에 따른 평균 성적')

    # x축명 변경
    axes[0].set_xticks([0, 1])
    axes[1].set_xticks([0, 1])

    # x, y축 라벨 설정
    axes[0].set_xlabel('sleep_sufficient')
    axes[0].set_ylabel('score_mean')

    axes[1].set_xlabel('phone_overuse')
    axes[1].set_ylabel('score_mean')

    plt.subplots_adjust(wspace=0.3) # subplot 간 간격 조정
    plt.tight_layout()
    
    to_path = os.path.join(os.path.dirname(BASE_DIR), 'outputs', 'group_comparison.png')
    plt.savefig(to_path)

    plt.close()

    print(f'\n수면 그룹 평균: \n{df_sleep}')
    print(f'\n폰 그룹 평균: \n{df_phone}')
    print(f'\n저장완료: {os.path.relpath(to_path, BASE_DIR)}') # 저장한 파일의 상대경로 보기
```

### 기능 6 - main() 함수로 전체 연결

```python
def main():
    df = load_clean_data(FILE_PATH)
    
    if df is None:
        return None
    
    plot_distribution(df)
    show_correlation(df)
    plot_scatter(df)
    plot_group_comparison(df)
    
    return df

if __name__ == "__main__":
    df = main()
```
